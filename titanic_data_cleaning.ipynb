# Titanic Dataset - Data Cleaning & Preprocessing
This notebook performs a complete, step-by-step data cleaning pipeline on the Titanic dataset using **pandas** and **NumPy**.

### Project Objectives:
- Load and inspect raw Titanic passenger data
- Identify and quantify missing values and duplicate records
- Impute missing values with appropriate statistical measures (median, mode, and constant placeholder)
- Clean, trim, and standardize string columns
- Convert columns to appropriate data types (`int`, `float`, `category`)
- Validate cleaning results and ensure zero missing values in target columns
- Export the final cleaned dataset to `data/titanic_cleaned.csv`


## Step 1: Import Libraries
Import `pandas` and `numpy` for data manipulation.

```python
import os
import pandas as pd
import numpy as np

print(f"Pandas version: {pd.__version__}")
print(f"NumPy version: {np.__version__}")
```

## Step 2: Load the Dataset
Load `data/Titanic-Dataset.csv` using pandas.

```python
# Resolve dataset path whether running from project root or notebooks/ directory
data_path = 'data/Titanic-Dataset.csv' if os.path.exists('data/Titanic-Dataset.csv') else '../data/Titanic-Dataset.csv'
df = pd.read_csv(data_path)
print(f"Dataset successfully loaded from: {data_path}")
```

## Step 3: Dataset Overview
Display the first 5 rows, dataset shape, column names, data types, and summary statistics.

```python
# Display first 5 rows
print("--- First 5 Rows ---")
df.head()
```

```python
# Dataset shape and column names
print(f"Dataset Shape: {df.shape[0]} rows, {df.shape[1]} columns\n")
print(f"Column Names:\n{list(df.columns)}")
```

```python
# Data types
print("--- Column Data Types ---")
df.dtypes
```

```python
# Summary statistics
print("--- Summary Statistics ---")
df.describe(include='all')
```

## Step 4: Identify Missing Values & Calculate Missing Percentages
Check missing values using `df.isnull().sum()` and compute missing percentages for all columns.

```python
# Count missing values per column
missing_count = df.isnull().sum()
missing_percent = (df.isnull().sum() / len(df)) * 100

missing_table = pd.DataFrame({
    'Missing Count': missing_count,
    'Missing Percentage (%)': missing_percent.round(2)
})
print("--- Missing Values Summary ---")
missing_table
```

## Step 5: Identify and Remove Duplicate Rows
Check duplicate rows using `df.duplicated().sum()` and drop any duplicate rows if present.

```python
# Check for duplicates
num_duplicates = df.duplicated().sum()
print(f"Total duplicate rows: {num_duplicates}")

if num_duplicates > 0:
    df = df.drop_duplicates()
    print(f"Removed duplicate rows. New shape: {df.shape}")
else:
    print("No duplicate rows found.")
```

## Step 6: Handle Missing Values
- Fill missing values in `Age` with the median age.
- Fill missing values in `Embarked` with the mode.
- Replace missing values in `Cabin` with `"Unknown"`.

```python
# Fill missing Age with median
median_age = df['Age'].median()
df['Age'] = df['Age'].fillna(median_age)
print(f"Imputed missing Age with median value: {median_age}")

# Fill missing Embarked with mode
mode_embarked = df['Embarked'].mode()[0]
df['Embarked'] = df['Embarked'].fillna(mode_embarked)
print(f"Imputed missing Embarked with mode value: '{mode_embarked}'")

# Replace missing Cabin with 'Unknown'
df['Cabin'] = df['Cabin'].fillna("Unknown")
print("Replaced missing Cabin entries with 'Unknown'")
```

## Step 7: Clean and Standardize Text Columns
- Remove leading and trailing spaces from text columns.
- Standardize `Sex` values to lowercase.
- Standardize `Cabin` values to uppercase.

```python
# Remove leading and trailing spaces from text columns
text_cols = df.select_dtypes(include=['object', 'string', 'str']).columns
for col in text_cols:
    df[col] = df[col].astype(str).str.strip()
print(f"Trimmed leading/trailing spaces from columns: {list(text_cols)}")

# Standardize Sex values to lowercase
df['Sex'] = df['Sex'].str.lower()
print("Standardized 'Sex' values to lowercase.")

# Standardize Cabin values to uppercase
df['Cabin'] = df['Cabin'].str.upper()
print("Standardized 'Cabin' values to uppercase.")
```

## Step 8: Correct Data Types
- Convert `PassengerId`, `Survived`, `Pclass`, `SibSp`, and `Parch` to integer type.
- Convert `Age` and `Fare` to float type.
- Convert `Sex` and `Embarked` to category type.

```python
# Convert specified columns to integer
int_cols = ['PassengerId', 'Survived', 'Pclass', 'SibSp', 'Parch']
for col in int_cols:
    df[col] = df[col].astype(int)

# Convert specified columns to float
float_cols = ['Age', 'Fare']
for col in float_cols:
    df[col] = df[col].astype(float)

# Convert specified columns to category
cat_cols = ['Sex', 'Embarked']
for col in cat_cols:
    df[col] = df[col].astype('category')

print("--- Updated Data Types ---")
df.dtypes
```

## Step 9: Check Unique Values
Verify unique values in `Sex`, `Embarked`, and `Pclass`.

```python
print(f"Unique values in Sex: {list(df['Sex'].unique())}")
print(f"Unique values in Embarked: {list(df['Embarked'].unique())}")
print(f"Unique values in Pclass: {list(df['Pclass'].unique())}")
```

## Step 10: Validate Missing Values in Age and Embarked
Confirm that no null values remain in `Age` and `Embarked`.

```python
remaining_age_nulls = df['Age'].isnull().sum()
remaining_embarked_nulls = df['Embarked'].isnull().sum()

print(f"Remaining nulls in Age: {remaining_age_nulls}")
print(f"Remaining nulls in Embarked: {remaining_embarked_nulls}")

assert remaining_age_nulls == 0, "Validation Error: Age still contains missing values!"
assert remaining_embarked_nulls == 0, "Validation Error: Embarked still contains missing values!"
print("Validation Successful: Zero missing values remain in Age and Embarked columns.")
```

## Step 11: Save the Cleaned Dataset
Export the cleaned dataframe to `data/titanic_cleaned.csv` with `index=False`.

```python
output_path = 'data/titanic_cleaned.csv' if os.path.exists('data') else '../data/titanic_cleaned.csv'
df.to_csv(output_path, index=False)
print(f"Cleaned dataset successfully saved to: {output_path}")
print(f"Final dataset dimensions: {df.shape}")
print("\nFirst 5 rows of cleaned data:")
df.head()
```
